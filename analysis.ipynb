```python
#Check for nullls and fill them
df[['win_count','response_count']]=df[['win_count','response_count']].fillna(0)
```

```python
#calculatte win rate per row
df['win_rate']=df['win_count'] 
df['response_count']
df['win_rate']=df['win_rate'].replace([float('inf'),-float('inf')],0)
```

```python
#biddar with highest win rate
bidder = df['win_count'].idxmax()
top_rate = df['win_count'].max()
print(f"The bidder with higher win rate is {top_rate} with win count is {bidder}")
```

```python
df = df[df['bidder'].astype(str).str.strip().astype(bool)]

# ------------------------------
# Top 5 bidders by frequency
top_5_bidders = df['bidder'].value_counts().head(5)
top_5_bidders_df = top_5_bidders.reset_index()
top_5_bidders_df.columns = ['Bidder', 'Count']
print("Top 5 Bidders by Frequency:")
print(top_5_bidders_df)
```
